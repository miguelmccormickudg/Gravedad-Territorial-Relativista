```python
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN
from pptx.dml.color import RGBColor

# Crear presentación
prs = Presentation()

# Color de fondo gris claro
background_color = RGBColor(230,230,230)

# Estilo de título
def add_title_slide(title, subtitle):
    slide_layout = prs.slide_layouts[0]
    slide = prs.slides.add_slide(slide_layout)
    slide.background.fill.solid()
    slide.background.fill.fore_color.rgb = background_color
    title_placeholder = slide.shapes.title
    subtitle_placeholder = slide.placeholders[1]
    title_placeholder.text = title
    subtitle_placeholder.text = subtitle

# Slide de contenido genérico
def add_content_slide(title, content):
    slide_layout = prs.slide_layouts[1]
    slide = prs.slides.add_slide(slide_layout)
    slide.background.fill.solid()
    slide.background.fill.fore_color.rgb = background_color
    title_placeholder = slide.shapes.title
    content_placeholder = slide.placeholders[1]
    title_placeholder.text = title
    tf = content_placeholder.text_frame
    for paragraph in content:
        p = tf.add_paragraph()
        p.text = paragraph
        p.space_after = Pt(10)

# Slide de tabla
def add_table_slide(title, headers, rows):
    slide_layout = prs.slide_layouts[5]
    slide = prs.slides.add_slide(slide_layout)
    slide.background.fill.solid()
    slide.background.fill.fore_color.rgb = background_color
    title_shape = slide.shapes.title
    title_shape.text = title

    rows_count = len(rows) + 1
    cols_count = len(headers)
    table = slide.shapes.add_table(rows_count, cols_count, Inches(0.5), Inches(1.5), Inches(9), Inches(4)).table

    # Encabezados
    for i, header in enumerate(headers):
        cell = table.cell(0, i)
        cell.text = header

    # Filas
    for row_idx, row in enumerate(rows):
        for col_idx, value in enumerate(row):
            table.cell(row_idx + 1, col_idx).text = value

# Slides

add_title_slide(
    "IA Generativa en la Educación Superior Pública",
    "Crear y transformar textos con recursos limitados"
)

add_content_slide(
    "¿Por qué hablar de IA generativa?",
    [
        "• La IA generativa ya forma parte del mundo académico.",
        "• Permite ahorrar tiempo en la creación de materiales.",
        "• Ayuda a personalizar contenidos según cada público.",
        "• No requiere infraestructura compleja."
    ]
)

add_content_slide(
    "¿Qué es la IA generativa?",
    [
        "• Herramienta que crea contenido original a partir de instrucciones.",
        "• Aprende patrones del lenguaje.",
        "• Puede crear, refinar y transformar textos."
    ]
)

add_content_slide(
    "¿Dónde se usa?",
    [
        "• Google Workspace (Duet AI): Gmail, Documentos.",
        "• ChatGPT (OpenAI): Navegador o aplicación."
    ]
)

add_content_slide(
    "Demostración en vivo",
    [
        "Ejemplo real:",
        "1. Prompt: Redacta un correo breve invitando a una reunión con estudiantes.",
        "2. Ajuste de tono: formal / cercano / motivador.",
        "3. Cambio de público: estudiantes / padres / colegas."
    ]
)

add_table_slide(
    "Comparativa de herramientas",
    ["Criterio", "ChatGPT", "Duet AI (Google)"],
    [
        ["Plataforma", "Navegador, App", "Gmail, Docs"],
        ["Registro", "Cuenta OpenAI", "Cuenta Google Workspace"],
        ["Idioma", "Español disponible", "Inglés y gradual en español"],
        ["Personalización", "Vía prompt detallado", "Vía opciones guiadas"],
        ["Costo", "Gratuito limitado", "Incluido en algunas cuentas"]
    ]
)

add_content_slide(
    "Caso práctico educativo",
    [
        "Prompt:",
        "Crea un texto breve que explique una actividad de investigación a estudiantes de ingeniería.",
        "Resultados mostrados:",
        "- ChatGPT: versión 1",
        "- Duet AI: versión 2"
    ]
)

add_content_slide(
    "Taller práctico",
    [
        "Instrucciones:",
        "1. Elige un texto breve que ya tengas.",
        "2. Genera 2 versiones con IA:",
        "   - Cambia el tono.",
        "   - Cambia la audiencia.",
        "3. Reflexiona: ¿Qué versión comunica mejor?"
    ]
)

add_content_slide(
    "Reflexión en grupos",
    [
        "Preguntas:",
        "• ¿Qué versión resultó más útil?",
        "• ¿Qué retos encontraste?",
        "• ¿Qué límites éticos identificaste?"
    ]
)

add_content_slide(
    "Beneficios en universidades públicas",
    [
        "• Ahorra tiempo al crear materiales.",
        "• Permite personalizar mensajes según el público.",
        "• Fomenta la innovación docente.",
        "• Requiere recursos mínimos."
    ]
)

add_content_slide(
    "Consideraciones éticas",
    [
        "Preguntas clave:",
        "• ¿Es legítimo presentar el contenido como propio?",
        "• ¿Cuándo citar la herramienta?",
        "• ¿Qué pasa con los datos sensibles?"
    ]
)

add_content_slide(
    "Cierre y próximos pasos",
    [
        "• Explora las herramientas con calma.",
        "• Comparte con colegas lo que aprendas.",
        "• Próxima sesión: Guerra de Prompts.",
        "La innovación no depende del presupuesto, sino de la disposición al cambio."
    ]
)

# Guardar presentación
prs.save("IA_Generativa_Educacion_Superior.pptx")
```
